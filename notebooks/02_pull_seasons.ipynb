```python
import pandas as pd
import time
```

```python
from nba_api.stats.endpoints import leaguegamefinder
```

```python
seasons = ['2021-22', '2022-23', '2023-24', '2024-25', '2025-26']
```

```python
all_seasons = []
```

```python
for season in seasons:
    finder = leaguegamefinder.LeagueGameFinder(season_nullable=season)
    game_df = finder.get_data_frames()[0]
    all_seasons.append(game_df)
    time.sleep(1)
```

```python
len(all_seasons)
```

```python
all_games = pd.concat(all_seasons, ignore_index=True)
```

```python
all_games.shape
```

```python
all_reg = all_games[all_games['SEASON_ID'].str.startswith('2')]
```

```python
all_reg.shape
```

```python
all_reg.to_csv('../data/games_2021_26.csv', index=False)
```
